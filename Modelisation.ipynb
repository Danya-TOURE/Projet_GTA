```python
import pandas as pd

# Charger les données dans cette cellule pour qu'elle puisse être exécutée seule.
df = pd.read_csv('gta_v_worldwide_sales_player_analytics_2013_2026.csv')

# Afficher le nombre de valeurs manquantes par colonne
print(df.isnull().sum())
```

```python
# Identifier les colonnes numériques et textuelles
colonnes_numeriques = df.select_dtypes(include=np.number).columns
colonnes_textuelles = df.select_dtypes(include=['object']).columns

# Remplacer les valeurs manquantes numériques par la médiane
for col in colonnes_numeriques:
    df[col] = df[col].fillna(df[col].median())

# Remplacer les valeurs manquantes textuelles par la valeur fréquente
for col in colonnes_textuelles:
    if df[col].isnull().any():
        df[col] = df[col].fillna(df[col].mode()[0])
```

```python
# Encoder les colonnes catégorielles
df = pd.get_dummies(
    df,
    columns=colonnes_textuelles,
    dtype=int
)

print(df.head())
```

```python
# Supprimer les colonnes identifiantes
colonnes_a_supprimer = ['ID']

df = df.drop(
    columns=colonnes_a_supprimer,
    errors='ignore'
)
```

```python
# Vérifier les valeurs manquantes
print("Valeurs manquantes :")
print(df.isnull().sum().sum())

# Vérifier les types de données
print("\nTypes des colonnes :")
print(df.dtypes)

# Vérifier que toutes les colonnes sont numériques
print("\nToutes les colonnes sont numériques :")
print(all(df.dtypes.apply(
    lambda x: np.issubdtype(x, np.number)
)))

# Afficher le dataset nettoyé
print(df.head())
```
